## Population Analysis

In this section, we analyze population-related trends across different countries.

```python
#which country has the higest population
df[df["population"] == df["population"].max()]["country"]
```

```python
# what is the capital of the country with higest population 
df[df["population"] == df["population"].max()]["capital_city"]
```

```python
#which country has the least population
df[df["population"] == df["population"].min()]["country"]
```

```python
#what is the capital of the country with least population
df[df["population"] == df["population"].min()]["capital_city"]
```

```python
#who is the political leader of the 2nd highest populated country
df[df["population"] == df["population"].nlargest(2).iloc[1]]["political_leader"]
```

```python
#which country in african region has highest population
africa_df = df[df["continent"] == "Africa"]
```

```python
#Top 10 Most Populated Countries
top10 = df.sort_values( by = 'population', ascending= False). head(10)
top10
```

```python
plt.figure(figsize= (10,5))
sns.barplot(data = top10 , y = 'country', x = 'population')
plt.title ('Top 10 Most Populated Countries')
plt.xlabel('Population')
plt.ylabel('Country')
```

Observation : Here we Notice that in the top 10 populated countries India and China and at the top.

## Political Analysis

This section explores democracy scores, political leaders, and governance-related information.

```python
#give me top 5 countries with highest democratic score
df.columns
```

```python
top_democracy = df.sort_values(by="democracy_score", ascending=False)
top_democracy.head()
```

```python
df['country'].head()
```

```python
# how many countries are there whoes political leaders are unknown
df[df["political_leader"].isnull()]["country"].count()
```

```python
#Average Democracy Score by Continent
avg_democracy = df.groupby('continent')['democracy_score'].mean().sort_values(ascending= False)
avg_democracy
```

```python
sns.barplot(x=avg_democracy.values, y=avg_democracy.index)
plt.title("Average Democracy Score by Continent")
plt.xlabel("Average Democracy Score")
plt.ylabel("Continent")
```

Observation : After Calculation of Average democracy score we can conclude that Europe and America has the highest average democracy score.
