# Team Performance Analysis

Which team won the most matches?

```python
match_wins = df['match_winner'].value_counts()
sns.barplot(y = match_wins.index , x = match_wins.values, palette = 'viridis')
plt.title("Number of Matches Won By Team")
```

**Observation:** Gujarat Titans won the highest number of matches during the IPL 2022 season.

How do teams win? (Runs vs Wickets)

```python
sns.countplot(x=df['won_by'])
plt.title('Won By')
```

**Observation:** The teams won equally with wickets and runs 

# Toss Analysis

Toss Decision Trends

```python
sns.countplot(x = df['toss_decision'], palette= 'rainbow')
plt.title("Toss Decisions")
```

**Observation:** We can observe that most of the teams chose to Field instead of Bat

Toss Winner vs Match Winner Percentage

```python
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
percentage = count*100/df.shape[0]
percentage.round(2)
```

# Player Performance Analysis

Key Player Performances

1 Most "Player of the Match" Awards

```python
count = df['player_of_the_match'].value_counts().head(10)
sns.barplot(x = count.values , y = count.index, palette= 'mako')
plt.title('Top 10 Players With Man Of The Match')
```

**Observation:** The Player of the Match awards were concentrated among a few consistently high-performing players, which were Kuldeep Yadav and Jos Buttler

2 Top Scorers

```python
df.head()
```

```python
high = df.groupby('top_scorer')['highscore'].max().sort_values(ascending = False).head(2)
high
```

```python
high.plot(kind = "bar")
```

**Observation:** The Player who scored the most number of runs across all the matches have been identified and their total scores across the matches have been calculated.

```python
df.head()
```

10 Best Bowling Figures

```python
df['higest_wickets'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['higest_wickets'] = df['higest_wickets'].astype(int)
top_bowlers = df.groupby('best_bowling')['higest_wickets'].sum().sort_values(ascending= False).head(10)
top_bowlers
```

```python
top_bowlers.plot(kind ='barh')
```

**Observation:** The Player who performed very well in bowling as we can observe are Yuzvendra Chahal and Kuldeep Yadav
