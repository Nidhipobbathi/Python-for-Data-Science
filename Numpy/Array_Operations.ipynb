DEEP AND SHALLOW COPY

```python
a = np.arange(1,21)
a
```

```python
slice = a[0:5]
slice = slice*10
slice
```

```python
a[4:6] *10
```

```python
a # so as we can see there are no changes in the sliced part of the original array , this is the concept of shallow copy where the changes made in the sliced part does not affect the actual array
```

so when you slice a portion of the array then that is stored in a different location.
so the changes made in the sliced portion will not affect the actual array.
but if we take the array and we make a copy of it and make chnages then the original array
will be modified. if we do not want modification then we do the following

```python
a 
b = a
```

```python
b[0]=23
b
```

```python
a #here we see the original array is changed 
```

```python
c = np.arange(1,21)
c
```

```python
c
d= c.copy()
```

```python
d[0]=23
d
```

```python
c # here our original array remains the same
```
